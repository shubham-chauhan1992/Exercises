## Create scope
databricks secrets create-scope <scope-name>
## delete scope
databricks secrets list-scopes
## create secret inside a scope
databricks secrets put-secret --json '{
  "scope": "<scope-name>",
  "key": "<key-name>",
  "string_value": "<secret>"
}'



```python
%python
# Define variables for storage and mount point
storage_account_name = "onprem2cloud2025"
container_name = "landingzone2025"
mount_point = "/mnt/adls"

# Get Service Principal credentials from Databricks Secret Scope
client_id = dbutils.secrets.get(scope="con2adls", key="client_id")
client_secret = dbutils.secrets.get(scope="con2adls", key="secret")
tenant_id = dbutils.secrets.get(scope="con2adls", key="tenant_id")

configs = {
    "fs.azure.account.auth.type": "OAuth",
    "fs.azure.account.oauth.provider.type": "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider",
    "fs.azure.account.oauth2.client.id": client_id,
    "fs.azure.account.oauth2.client.secret": client_secret,
    "fs.azure.account.oauth2.client.endpoint": f"https://login.microsoftonline.com/{tenant_id}/oauth2/token"
}

dbutils.fs.mount(
    source=f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/",
    mount_point=mount_point,
    extra_configs=configs
)
```
